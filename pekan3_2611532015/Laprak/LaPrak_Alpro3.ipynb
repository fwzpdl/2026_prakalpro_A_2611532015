## 2.1 Dasar Teori

1. Operator dalam Python<br>
Operator adalah simbol khusus yang digunakan untuk melakukan suatu operasi terhadap satu atau lebih nilai (operand) dan menghasilkan suatu nilai baru sebagai hasilnya. Sebagai contoh, pada ekspresi 5 + 6, tanda + merupakan operator yang melakukan operasi penjumlahan terhadap operand 5 dan 6 yang mengembalikan hasil penjumlahan dari kedua operand tersebut. Python sendiri membagi operator ke dalam beberapa kelompok, yaitu operator aritmatika, assignment, perbandingan, logika, identitas, keanggotaan, dan bitwise

2. Operator Aritmatika <br>
Operator aritmatika digunakan untuk melakukan operasi matematika dasar seperti penjumlahan, pengurangan, perkalian, dan pembagian pada nilai numerik yang secara umum berfungsi untuk melakukan perhitungan matematis pada nilai numerik. Operator yang termasuk dalam kelompok ini antara lain + (tambah), - (kurang), * (kali), / (bagi), // (bagi bulat ke bawah), % (sisa bagi/modulus), dan ** (pangkat) di mana urutan pengerjaannya mengikuti kaidah PEMDAS seperti pada operasi matematika biasa.
3. Operator Assignment <br>
Operator assignment digunakan untuk memberikan atau memperbarui nilai suatu variabel dan biasanya dikombinasikan dengan operator lain seperti aritmatika, di mana operasi dilakukan lebih dulu kemudian hasilnya disimpan ke operand di sebelah kiri. Contohnya adalah = untuk memberi nilai awal, serta bentuk gabungan seperti +=, -=, *=, /=, dan %= yang berfungsi sebagai bentuk singkat dari operasi disertai penugasan ulang ke variabel yang sama.
4. Operator Perbandingan <br>
Operator perbandingan digunakan untuk membandingkan dua nilai dan selalu mengembalikan hasil berupa nilai boolean (True atau False). Operator yang termasuk kelompok ini adalah == (sama dengan), != (tidak sama dengan), > (lebih besar), < (lebih kecil), >= (lebih besar sama dengan), dan <= (lebih kecil sama dengan).
5. Operator Logika <br>
Operator logika digunakan untuk menggabungkan dua atau lebih ekspresi boolean menjadi satu kondisi dan secara umum dapat diterapkan pada berbagai jenis objek untuk mendukung pengujian kebenaran (truth test). Operator yang termasuk dalam kelompok ini adalah and (dan), or (atau), dan not (bukan/negasi).
6. Operator Identitas <br>
Operator identitas digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama persis di memori, bukan sekadar memiliki nilai yang sama yaitu dengan memeriksa apakah kedua objek tersebut memiliki nilai id yang sama, yang berarti keduanya menunjuk ke lokasi memori yang sama. Operator ini terdiri dari is dan is not.
7. Operator Keanggotaan <br>
Operator keanggotaan digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah struktur data seperti list, tuple, atau string yang menguji apakah suatu urutan (sequence) memiliki item tertentu atau tidak. Operator ini terdiri dari in dan not in.
8. Operator Bitwise <br>
Operator bitwise digunakan untuk melakukan operasi pada level bit (representasi biner) dari suatu bilangan di mana operasi ini dilakukan langsung pada operand dalam bentuk biner. Operator yang termasuk kelompok ini antara lain & (AND), | (OR), ^ (XOR), ~ (NOT), << (geser bit ke kiri), dan >> (geser bit ke kanan).

## 2.2 Kode Program

### 2.2.1 Program "A_2611532015"
**a. Operator Aritmatika**


```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_2015 = int(input("Input angka-1: "))
angka2_2015 = int(input("Input angka-2: "))

# Penjumlahan
hasil_2015 = angka1_2015 + angka2_2015
print("\nOperator Penjumlahan")
print("Hasil =",hasil_2015)

# Pengurangan
hasil_2015 = angka1_2015 - angka2_2015
print("\nOperator Pengurangan")
print("Hasil =",hasil_2015)

# Perkalian
hasil_2015 = angka1_2015 * angka2_2015
print("\nOperator Perkalian")
print("Hasil =",hasil_2015)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_2015 != 0:
    hasil_2015 = angka1_2015 / angka2_2015
    print("\nOperator Pembagian")
    print("Hasil =",hasil_2015)

    hasil_2015 = angka1_2015 // angka2_2015
    print("\nOperator Pembagian Bulat")
    print("Hasil =",hasil_2015)

    hasil_2015 = angka1_2015 % angka2_2015
    print("\nOperator Sisa Bagi")
    print("Hasil =",hasil_2015)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_2015 = angka1_2015 ** angka2_2015
print("\nOperator Pangkat")
print("Hasil =",hasil_2015)
```

**Penjelasan :**
1. angka1_2015 = int(input("Input angka-1: ")) dan angka2_2015 = int(input("Input angka-2: ")) — program meminta pengguna memasukkan dua buah angka, kemudian nilai yang dimasukkan (berupa teks) diubah menjadi tipe data integer menggunakan int() agar dapat diproses secara matematis.

2. hasil_2015 = angka1_2015 + angka2_2015, melakukan operasi penjumlahan antara angka1_2015 dan angka2_2015, kemudian hasilnya disimpan ke variabel hasil_2015 dan ditampilkan menggunakan print().

3. hasil_2015 = angka1_2015 - angka2_2015, melakukan operasi pengurangan, yaitu angka1_2015 dikurangi angka2_2015, hasilnya disimpan dan ditampilkan dengan cara yang sama.

4. hasil_2015 = angka1_2015 * angka2_2015, melakukan operasi perkalian antara kedua angka yang dimasukkan.

5. if angka2_2015 != 0:, merupakan struktur percabangan yang mengecek apakah angka2_2015 tidak sama dengan 0. Pengecekan ini diperlukan karena operasi pembagian dengan angka 0 tidak diperbolehkan dan akan menyebabkan program error.

6. hasil_2015 = angka1_2015 / angka2_2015, melakukan operasi pembagian biasa, hasilnya berupa bilangan pecahan (float).

7. hasil_2015 = angka1_2015 // angka2_2015, melakukan operasi pembagian bulat (floor division), yaitu hasil pembagian dibulatkan ke bawah menjadi bilangan bulat.

8. hasil_2015 = angka1_2015 % angka2_2015, melakukan operasi modulus, yaitu mengambil sisa hasil pembagian antara angka1_2015 dan angka2_2015.

9. else: print("Angka kedua tidak boleh bernilai 0."), bagian ini akan dijalankan apabila angka2_2015 bernilai 0, yaitu dengan menampilkan pesan peringatan agar operasi pembagian tidak dijalankan.

10. hasil_2015 = angka1_2015 ** angka2_2015, melakukan operasi pangkat, yaitu angka1_2015 dipangkatkan dengan angka2_2015.

**b. Operator Logika**

```python
#Program operator logika dalam python

#memasukkan nilai boolean
#input tidak peka terhadap huruf besar dan kecil

a1_2015 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_2015 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 = ", a1_2015)
print("A2 =", a2_2015)

#Konjungsi : bernilai true jika keduanya true
hasil_2015 = a1_2015 and a2_2015
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil_2015)

#disjungsi: bernilai true jika salah satunya true
hasil_2015 = a1_2015 or a2_2015
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil_2015)

#Negasi A1: membalik nilai a1
hasil_2015 = not a1_2015
print("\nNegasi A1 (NOt)")
print("not A1=", hasil_2015)

#Negasi A2: membalik nilai a2
hasil_2015 = not a2_2015
print("\nNegasi A2 (NOt)")
print("not A2=", hasil_2015)

# XOR: bernilai true jika kedua nilai berbeda
hasil_2015 = a1_2015 != a2_2015
print("\nDisjungsi ekslusif (XOR)")
print(" A1 XOR A2 =", hasil_2015)

```

**Penjelasan:**

Fungsi .strip() menghapus spasi di awal/akhir input, .lower() mengubah semua huruf menjadi huruf kecil agar tidak peka huruf besar/kecil, kemudian hasilnya dibandingkan dengan "true" menggunakan == sehingga a1_2015 bernilai True jika input sesuai, atau False jika tidak.

2. a2_2015 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true", melakukan proses yang sama seperti poin 1, namun untuk nilai boolean kedua yang disimpan ke variabel a2_2015.

3. hasil_2015 = a1_2015 and a2_2015, melakukan operasi konjungsi (AND), yaitu hasilnya bernilai True hanya jika a1_2015 dan a2_2015 sama-sama bernilai True.

4. hasil_2015 = a1_2015 or a2_2015, melakukan operasi disjungsi (OR), yaitu hasilnya bernilai True jika salah satu dari a1_2015 atau a2_2015 bernilai True.

5. hasil_2015 = not a1_2015, melakukan operasi negasi terhadap a1_2015, yaitu membalik nilainya (dari True menjadi False, atau sebaliknya).

6. hasil_2015 = not a2_2015, melakukan operasi negasi terhadap a2_2015 dengan cara yang sama seperti poin 5.

7. hasil_2015 = a1_2015 != a2_2015, melakukan operasi disjungsi eksklusif (XOR) dengan memanfaatkan operator !=, yaitu hasilnya bernilai True apabila a1_2015 dan a2_2015 memiliki nilai yang berbeda, dan bernilai False jika nilainya sama.

**c. Operator BitWise.**

```python


print("===============================")
print("3. OPERATOR BITWISE")
print("===============================")

angka1_2015 = int(input("Masukkan angka bitwise-1 :"))
angka2_2015 = int(input("Masukkan angka bitwise-2 :"))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_2015, "|biner =", bin(angka1_2015))
print("angka2 =", angka2_2015, "|biner =", bin(angka2_2015))

# Bitwise AND
hasil_2015 = angka1_2015 & angka2_2015
print("\nBitwise AND (&)")
print(angka1_2015, "&", angka2_2015, "=", hasil_2015)
print("Biner hasil =", bin(hasil_2015))
print("Biner hasil (8 bit) =", format(hasil_2015, "08b"))

# Bitwise OR
hasil_2015 = angka1_2015 | angka2_2015
print("\nBitwise OR(|)")
print(angka1_2015, "|", angka2_2015, "=", hasil_2015)
print("Biner hasil =", bin(hasil_2015))
print("Biner hasil (8 bit) =", format(hasil_2015, "08b"))

# Bitwise XOR
hasil_2015 = angka1_2015 ^ angka2_2015
print("\nBitwise XOR (^)")
print(angka1_2015, "^", angka2_2015, "=", hasil_2015)
print("Biner hasil =", bin(hasil_2015))
print("Biner hasil (8 bit) =", format(hasil_2015, "08b"))

# Bitwise NOT
hasil_2015 = ~angka1_2015
print("\nBitwise NOT (~)")
print("~", angka1_2015, "=", hasil_2015)
print("Biner hasil =", bin(hasil_2015))
print("Biner hasil (8 bit) =", format(hasil_2015, "08b"))

# Bitwise geser kiri
jumlah_geser_2015 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2015 = angka1_2015 << jumlah_geser_2015
print("\nBitwise geser kiri (<<)")
print(angka1_2015, "<<", jumlah_geser_2015, "=", hasil_2015)
print("Biner hasil =", bin(hasil_2015))
print("Biner hasil (8 bit) =", format(hasil_2015, "08b"))

# Bitwise geser kanan
hasil_2015 = angka1_2015 >> jumlah_geser_2015
print("\nBitwise geser kanan (>>)")
print(angka1_2015, ">>", jumlah_geser_2015, "=", hasil_2015)
print("Biner hasil =", bin(hasil_2015))
print("Biner hasil (8 bit) =", format(hasil_2015, "08b"))
```

**Penjelasan:**

1. angka1_2015 = int(input("Masukkan angka bitwise-1 :")) dan angka2_2015 = int(input("Masukkan angka bitwise-2 :")) — program meminta pengguna memasukkan dua buah angka, kemudian nilai yang dimasukkan diubah menjadi tipe data integer menggunakan int() agar dapat diproses dalam operasi bitwise.

2. print("angka1 =", angka1_2015, "|biner =", bin(angka1_2015)) dan baris sejenisnya untuk angka2_2015 — menampilkan nilai kedua angka dalam bentuk desimal sekaligus bentuk binernya menggunakan fungsi bin(), sehingga representasi bit dari masing-masing angka dapat dilihat sebelum diproses.

3. hasil_2015 = angka1_2015 & angka2_2015 — melakukan operasi bitwise AND, yaitu membandingkan setiap bit dari angka1_2015 dan angka2_2015, hasil bitnya akan bernilai 1 hanya jika kedua bit yang dibandingkan sama-sama bernilai 1.

4. hasil_2015 = angka1_2015 | angka2_2015 — melakukan operasi bitwise OR, yaitu hasil bitnya akan bernilai 1 apabila salah satu atau kedua bit yang dibandingkan bernilai 1.

5. hasil_2015 = angka1_2015 ^ angka2_2015 — melakukan operasi bitwise XOR, yaitu hasil bitnya akan bernilai 1 apabila kedua bit yang dibandingkan berbeda nilai, dan bernilai 0 apabila sama.

6. hasil_2015 = ~angka1_2015 — melakukan operasi bitwise NOT, yaitu membalik seluruh bit dari angka1_2015 (bit 0 menjadi 1 dan sebaliknya), yang secara matematis setara dengan -(angka1_2015 + 1).

7. jumlah_geser_2015 = int(input("\nMasukkan jumlah pergeseran bit: ")) — meminta pengguna memasukkan jumlah bit yang akan digeser, kemudian nilainya diubah menjadi tipe data integer.

8. hasil_2015 = angka1_2015 << jumlah_geser_2015 — melakukan operasi geser bit ke kiri sebanyak jumlah_geser_2015 posisi, yang secara efektif mengalikan nilai angka1_2015 dengan 2 pangkat jumlah pergeseran.

9. hasil_2015 = angka1_2015 >> jumlah_geser_2015 — melakukan operasi geser bit ke kanan sebanyak jumlah_geser_2015 posisi, yang secara efektif membagi nilai angka1_2015 dengan 2 pangkat jumlah pergeseran (dibulatkan ke bawah).

10. print("Biner hasil (8 bit) =", format(hasil_2015, "08b")) — menampilkan hasil setiap operasi dalam bentuk biner dengan format tetap 8 digit (menggunakan format() dan spesifier "08b"), sehingga tampilan biner lebih rapi dan mudah dibaca.

**d. Operator Perbandingan**


```python



angka1_2015 = int(input("Input angka ke-1 : "))
angka2_2015 = int(input("Input angka ke-2 : "))

#Lebih besar dari
hasil_2015 = angka1_2015 > angka2_2015
print("\nOperator lenih besar dari")
print("angka1_2015 > angka2_2015 = ", hasil_2015)

#Lebih kecil dari
hasil_2015 = angka1_2015 < angka2_2015
print("\nOperator lebih kecil dari")
print("angka1_2015 < angka2_2015 = ", hasil_2015)

#Lebih besar atau sama dengan 
hasil_2015 = angka1_2015 >= angka2_2015
print("\nOperator lebih kecil dari")
print("angka1_2015 >= angka2_2015 = ", hasil_2015)

#Lebih kecil atau sama dengan 
hasil_2015 = angka1_2015 <= angka2_2015
print("\nOperator lebih kecil dari")
print("angka1_2015 <= angka2_2015 = ", hasil_2015)

#Sama dengan
hasil_2015 = angka1_2015 == angka2_2015
print("\nOperator sama dengan")
print("angka1_2015 == angka2_2015 = ", hasil_2015)

#Tidak sama dengan
hasil_2015 = angka1_2015 != angka2_2015
print("\nOperator tidak sama dengan")
print("angka1_2015 != angka2_2015 = ", hasil_2015)

# Tambahan: perbandingan berantai dalam python
hasil_2015 = 0 < angka1_2015 < 100
print("\nPerbandingan berantai")
print("0 < angka1_2015 < 100 = ", hasil_2015)
```

2. hasil_2015 = angka1_2015 > angka2_2015 — melakukan operasi perbandingan "lebih besar dari", hasilnya bernilai True apabila angka1_2015 lebih besar dari angka2_2015, dan False jika sebaliknya.

3. hasil_2015 = angka1_2015 < angka2_2015 — melakukan operasi perbandingan "lebih kecil dari", hasilnya bernilai True apabila angka1_2015 lebih kecil dari angka2_2015.dengan fungsi int(), lalu menghitung perkalian panjang × lebar untuk mendapatkan nilai luas.

4. hasil_2015 = angka1_2015 >= angka2_2015 — melakukan operasi perbandingan "lebih besar atau sama dengan", hasilnya bernilai True apabila angka1_2015 lebih besar dari atau sama dengan angka2_2015.

5. hasil_2015 = angka1_2015 <= angka2_2015 — melakukan operasi perbandingan "lebih kecil atau sama dengan", hasilnya bernilai True apabila angka1_2015 lebih kecil dari atau sama dengan angka2_2015.

6. hasil_2015 = angka1_2015 == angka2_2015 — melakukan operasi perbandingan "sama dengan", hasilnya bernilai True apabila nilai angka1_2015 dan angka2_2015 sama persis.

**e. Operator Assignment**

```python
#program operator assignment dalam python

angka1_2015 = int(input("Input angka-1: "))
angka2_2015 = int(input("Input angka-2: "))

print("\nNilai awal angka1_2015 =", angka1_2015)
print("Nilai angka2_2015 =", angka2_2015)

# Assignmrnt biasa
hasil_2015 = angka1_2015
print("\nAssignment biasa (=)")
print("hasil =", hasil_2015)

# assignment penambahan
hasil_2015 = angka1_2015
hasil_2015 += angka2_2015
print("\nAssignment penjumlahan(+)")
print("hasil =", hasil_2015)

# assignment pengurangan
hasil_2015 = angka1_2015
hasil_2015 -= angka2_2015
print("\nAssignment pengurangan(-)")
print("hasil =", hasil_2015)

# assignment perkalian
hasil_2015 = angka1_2015
hasil_2015 *= angka2_2015
print("\nAssignment perkalian(*)")
print("hasil =", hasil_2015)

# assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_2015 != 0:
    hasil_2015 = angka1_2015
    hasil_2015 /= angka2_2015
    print("\nAssignment pembagian(/)")
    print("hasil =", hasil_2015)

    #operasi tambahan
    hasil_2015 = angka1_2015
    hasil_2015 //= angka2_2015
    print("\nAssignment pembagian bulat(//)")
    print("hasil =", hasil_2015)

    hasil_2015 = angka1_2015
    hasil_2015 %= angka2_2015
    print("\nAssignment sisa bagi (%)")
    print("hasil =", hasil_2015)
else:
    print("Pembagian tidak dapat dilakukan")
    print("Angka kedua tidak boleh 0")

# Operator tambahan : perpangkatan
hasil_2015 = angka1_2015 
hasil_2015 **= angka2_2015
print("\nOperator perpangkatan (**=)")
print("Hasil =",hasil_2015)
```

**Penjelasan:**

1. angka1_2015 = int(input("Input angka-1: ")) dan angka2_2015 = int(input("Input angka-2: ")) — program meminta pengguna memasukkan dua buah angka, kemudian nilai yang dimasukkan diubah menjadi tipe data integer menggunakan int().

2. hasil_2015 = angka1_2015 (Assignment biasa) — menggunakan operator = untuk memberikan nilai angka1_2015 ke variabel hasil_2015 secara langsung, tanpa operasi tambahan apapun.

3. hasil_2015 = angka1_2015 lalu hasil_2015 += angka2_2015 — nilai angka1_2015 disalin ke hasil_2015, kemudian operator += menjumlahkan hasil_2015 dengan angka2_2015 dan menyimpan hasilnya kembali ke hasil_2015, yang setara dengan hasil_2015 = hasil_2015 + angka2_2015.

4. hasil_2015 -= angka2_2015 — operator -= mengurangkan hasil_2015 dengan angka2_2015 dan menyimpan hasilnya kembali ke hasil_2015, yang setara dengan hasil_2015 = hasil_2015 - angka2_2015.

5. hasil_2015 *= angka2_2015 — operator *= mengalikan hasil_2015 dengan angka2_2015 dan menyimpan hasilnya kembali ke hasil_2015, yang setara dengan hasil_2015 = hasil_2015 * angka2_2015.

6. if angka2_2015 != 0: — struktur percabangan yang mengecek apakah angka2_2015 tidak sama dengan 0, untuk mencegah terjadinya error akibat pembagian dengan nilai 0 pada operasi-operasi berikutnya.

7. hasil_2015 /= angka2_2015 — operator /= membagi hasil_2015 dengan angka2_2015 dan menyimpan hasilnya (berupa bilangan pecahan/float) kembali ke hasil_2015.

8. hasil_2015 //= angka2_2015 — operator //= membagi hasil_2015 dengan angka2_2015 secara pembagian bulat (floor division), hasilnya dibulatkan ke bawah dan disimpan kembali ke hasil_2015.

9. hasil_2015 %= angka2_2015 — operator %= mengambil sisa hasil pembagian hasil_2015 dengan angka2_2015, kemudian hasilnya disimpan kembali ke hasil_2015.

10. else: print(...) — bagian ini akan dijalankan apabila angka2_2015 bernilai 0, yaitu dengan menampilkan pesan bahwa operasi pembagian tidak dapat dilakukan.

11. hasil_2015 **= angka2_2015 — operator **= memangkatkan hasil_2015 dengan angka2_2015 dan menyimpan hasilnya kembali ke hasil_2015, yang setara dengan hasil_2015 = hasil_2015 ** angka2_2015.

**f. Operator Keanggotaan dan Identitas**

```python


print("===============================")
print("1. OPERATOR KEANGGOTAAN")
print("===============================")

#INPUT BEBERAPA DATA YANG DIPISAHKAN KOMA
input_data_2015 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list ineteger
data_2015 = [int(angka.strip()) for angka in input_data_2015.split(",")]

nilai_dicari_2015 = int(input("Masukkan angka yang ingin dicari: "))

#Operator in
hasil_2015 = nilai_dicari_2015 in data_2015
print("\nOperator Keanggotaan IN")
print(nilai_dicari_2015, "in", data_2015, "=", hasil_2015)

#Operator not in 
hasil_2015 = nilai_dicari_2015 not in data_2015
print("\nOperator Keanggotaan NOT IN")
print(nilai_dicari_2015, "not in", data_2015, "=", hasil_2015)



print("===============================")
print("2. OPERATOR IDENTITAS")
print("===============================")

# objek1 menggunakan list dari input pengguna
objek1_2015 = data_2015

# objek2 merujuk pada objek yang sama dengan objek1
objek2_2015 = objek1_2015

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_2015 = data_2015.copy()

print("objek1 =", objek1_2015)
print("objek2 =", objek2_2015)
print("objek3 =", objek3_2015)

# OPERATOR IS
hasil_2015 = objek1_2015 is objek2_2015
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil_2015)

#Operator Is Not
hasil_2015 = objek1_2015 is not objek2_2015
print("\nOperator identitas IS NOT")
print("objek1 is not objek2 =", hasil_2015)

#membandingkan identitas dan nilai
print("\nPerbandingan Identitas dan Nilai")
print("objek1 is objek3 =", objek1_2015 is objek3_2015 )
print("objek1 == objek3 =", objek1_2015 == objek3_2015)
```

**Penjelasan:**

1. input_data_2015 = input("Masukkan beberapa angka, pisahkan dengan koma: ") — program meminta pengguna memasukkan beberapa angka sekaligus dalam satu baris, dengan setiap angka dipisahkan oleh tanda koma, hasil input ditampung dalam bentuk teks (string).

2. data_2015 = [int(angka.strip()) for angka in input_data_2015.split(",")] — teks input dipecah menjadi beberapa bagian berdasarkan tanda koma menggunakan .split(","), kemudian setiap bagian dihilangkan spasinya dengan .strip() dan diubah menjadi tipe data integer menggunakan int(). Proses ini dilakukan sekaligus menggunakan list comprehension, sehingga hasil akhirnya berupa list berisi angka-angka integer.

3. nilai_dicari_2015 = int(input("Masukkan angka yang ingin dicari: ")) — program meminta pengguna memasukkan satu angka yang akan dicari keberadaannya di dalam list, kemudian nilainya diubah menjadi tipe data integer.

4. hasil_2015 = nilai_dicari_2015 in data_2015 — menggunakan operator in untuk memeriksa apakah nilai_dicari_2015 terdapat di dalam list data_2015, hasilnya bernilai True jika angka tersebut ditemukan, dan False jika tidak ditemukan.

5. hasil_2015 = nilai_dicari_2015 not in data_2015 — menggunakan operator not in untuk memeriksa kebalikan dari poin sebelumnya, yaitu hasilnya bernilai True apabila nilai_dicari_2015 tidak terdapat di dalam list data_2015, dan False apabila angka tersebut ditemukan.

6. objek1_2015 = data_2015 — variabel objek1_2015 dibuat dengan merujuk pada list data_2015 yang sudah diisi sebelumnya, sehingga objek1_2015 menunjuk ke objek list yang sama dengan data_2015.

7. objek2_2015 = objek1_2015 — variabel objek2_2015 dibuat dengan menyalin referensi dari objek1_2015, sehingga objek2_2015 bukan membuat list baru, melainkan hanya menunjuk ke objek list yang sama persis di memori dengan objek1_2015.

8. objek3_2015 = data_2015.copy() — variabel objek3_2015 dibuat menggunakan method .copy(), sehingga isi list-nya sama dengan data_2015, namun objek3_2015 merupakan objek list baru yang berdiri sendiri di memori, berbeda dengan data_2015 maupun objek1_2015.

9. hasil_2015 = objek1_2015 is objek2_2015 — menggunakan operator is untuk memeriksa apakah objek1_2015 dan objek2_2015 menunjuk ke objek yang sama persis di memori, hasilnya bernilai True karena keduanya memang merujuk ke objek yang sama (sesuai nomor 7)

10. hasil_2015 = objek1_2015 is not objek2_2015 — menggunakan operator is not untuk memeriksa kebalikan dari poin sebelumnya, hasilnya bernilai False karena objek1_2015 dan objek2_2015 memang menunjuk ke objek yang sama.

11. print("objek1 is objek3 =", objek1_2015 is objek3_2015) — memeriksa apakah objek1_2015 dan objek3_2015 menunjuk ke objek yang sama di memori, hasilnya bernilai False karena objek3_2015 merupakan objek baru hasil .copy(), meskipun isinya sama.

12. print("objek1 == objek3 =", objek1_2015 == objek3_2015) — memeriksa apakah nilai/isi dari objek1_2015 dan objek3_2015 sama menggunakan operator ==, hasilnya bernilai True karena kedua list tersebut memiliki elemen yang identik, meskipun merupakan objek yang berbeda di memori — inilah yang menjadi perbedaan mendasar antara operator identitas (is) dan operator perbandingan nilai (==).

## Tugas Praktikum
**Soal:**<br>
**Studi Kasus: Sistem Simulasi Transaksi dan Validasi Akses Toko**<br>
Anda diminta untuk membuat sebuah program kasir sederhana yang digunakan untuk menghitung total transaksi pelanggan sekaligus melakukan pemeriksaan terhadap status pelanggan dan hak akses terhadap promo tertentu. Program menerima data transaksi dan informasi pelanggan, kemudian menggunakan berbagai jenis operator Python untuk menentukan total pembayaran, status diskon, kelayakan promo, serta hak akses pelanggan. Seluruh proses harus dibuat dalam satu alur program yang utuh.



**Kode Program:**

```python
# SISTEM OPERASI TOKO
nama_2015 = input("Masukkan Nama Pelanggan: ")
status_2015 = input("Masukkan Status Pelanggan (member/nonmember): ")
total_belanja_2015 = int(input("Masukkan total belanja: "))
jumlah_barang_2015 = int(input("Masukkan jumlah barang: "))
kode_promo_2015 = input("Masukkan Kode promo : ")

# Daftar promo tersedia
promo_list_2015 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]

#OPERATOR PERBANDINGAN
syarat_belanja_2015 = total_belanja_2015 >= 200000
syarat_barang_2015 = jumlah_barang_2015 >= 3
status_member_2015 = status_2015 == "member"
promo_tersedia_2015 = kode_promo_2015 in promo_list_2015

#Operator Logika 
diskon_member_2015 = status_member_2015 and syarat_belanja_2015
promo_dapat_2015 = syarat_barang_2015 or promo_tersedia_2015

#Operator Keanggotaan
cek_promo_2015 = kode_promo_2015 in promo_list_2015
cek_promo_invalid_2015 = kode_promo_2015 not in promo_list_2015

#Operator Aritmatika
diskon_2015 = 0
if diskon_member_2015:
    diskon_2015 = total_belanja_2015 * 10 // 100 #diskon 10%

#operator penugasan
total_bayar_2015 = total_belanja_2015
total_bayar_2015 -= diskon_2015

rata_barang_2015 = total_bayar_2015 / jumlah_barang_2015
sisa_pembagian_2015 = total_bayar_2015 % jumlah_barang_2015

#operator identitas
angka_a_2015 = [1, 2, 3]
angka_b_2015 = [1, 2, 3]
identitas_sama_2015 = angka_a_2015 is angka_b_2015
nilai_sama_2015 = angka_a_2015 == angka_b_2015

#operator bitwise
kode_status_2015 = 0
if status_member_2015: kode_status_2015 |= 0b0001
if syarat_belanja_2015: kode_status_2015 |= 0b0010
if syarat_barang_2015: kode_status_2015 |= 0b0100
if promo_tersedia_2015: kode_status_2015 |= 0b1000

#Pemeriksaan bitwise
cek_member_bit_2015 = kode_status_2015 & 0b0001
cek_promo_bit_2015 = kode_status_2015 & 0b1000
perbandingan_bit_2015 = kode_status_2015 ^ 0b1011
shift_bit_2015 = kode_status_2015 << 1

#OUTPUT
print("\n=== DATA TRANSAKSI ===")
print("Nama Pelanggan       :", nama_2015)
print("Status Pelanggan     :", status_2015)
print("Total Belanja        : Rp", total_belanja_2015)
print("Jumlah Barang        :", jumlah_barang_2015)
print("Kode Promo           :", kode_promo_2015)

print("\n=== HASIL VALIDASI ===")
print("Belanja >= Rp200000        :", syarat_belanja_2015)
print("Jumlah Barang >= 3         :", syarat_barang_2015)
print("Status Member              :", status_member_2015)
print("Kode Promo Tersedia        :", promo_tersedia_2015)
print("Mendapatkan Diskon         :", diskon_member_2015)
print("Mendapatkan Promo          :", promo_dapat_2015)

print("\n=== HASIL PERHITUNGAN ===")
print("Diskon                     : Rp", diskon_2015)
print("Total Pembayaran           : Rp", total_bayar_2015)
print("Rata-rata Harga Barang     : Rp", int(rata_barang_2015))
print("Sisa Pembagian             :", sisa_pembagian_2015)

print("\n=== HASIL OPERATOR IDENTITAS ===")
print("Apakah identitas sama?     :", identitas_sama_2015)
print("Apakah nilai sama?         :", nilai_sama_2015)

print("\n=== HASIL BITWISE ===")
print("Kode Status (biner)        :", format(kode_status_2015, '04b'))
print("Kode Status (desimal)      :", kode_status_2015)
print("Cek Member (bitwise AND)   :", cek_member_bit_2015)
print("Cek Promo (bitwise AND)    :", cek_promo_bit_2015)
print("Perbandingan XOR           :", perbandingan_bit_2015)
print("Shift Left                 :", shift_bit_2015)

print("\n=== SELESAI ===")
```

    <br>and → menghasilkan True hanya jika kedua kondisi benar (contoh: pelanggan member dan belanja ≥ Rp200.000).
