## 2.1 Dasar Teori

1. Perulangan (Looping)<br>
Perulangan adalah struktur kontrol yang memungkinkan sekumpulan instruksi dijalankan berulang kali selama syarat tertentu terpenuhi atau sebanyak jumlah yang ditentukan. Dengan perulangan, kode program menjadi lebih singkat, efisien, dan mudah dipelihara karena tidak perlu menulis instruksi yang sama berkali-kali.
2. Perulangan for pada Python <br>
Perulangan for pada Python bekerja dengan menelusuri anggota dari suatu rangkaian nilai (sequence), lalu menjalankan blok kode di dalamnya satu kali untuk setiap nilai.
3.Fungsi range()<br>
Fungsi range() menghasilkan rangkaian bilangan bulat yang digunakan sebagai pengatur perulangan for. Fungsi ini memiliki tiga bentuk penulisan:<br>
-range(stop): dimulai dari 0 dan berhenti sebelum nilai stop.<br>
-range(start, stop): dimulai dari start dan berhenti sebelum nilai stop.<br>
-range(start, stop, step): dimulai dari start, berhenti sebelum stop, dengan kenaikan atau penurunan sebesar step.
Nilai stop tidak ikut dihitung. Karena itu, untuk menghasilkan angka 1 sampai N ditulis range(1, N + 1). Nilai step negatif digunakan untuk perulangan menurun, misalnya range(5, 0, -1) menghasilkan 5, 4, 3, 2, 1.
4. Perulangan Bersarang (Nested Loop)<br>
Nested loop adalah perulangan yang berada di dalam perulangan lain. Setiap kali perulangan luar berjalan satu kali, perulangan dalam dijalankan sampai selesai. Jika perulangan luar berjalan m kali dan perulangan dalam berjalan n kali, maka blok kode terdalam dijalankan sebanyak m × n kali.
5. Peran Perulangan Luar dan Dalam pada Pembentukan Pola
Dalam pencetakan pola dua dimensi, perulangan luar umumnya mengatur baris, sedangkan perulangan dalam mengatur kolom atau isi setiap baris. Jumlah ulangan perulangan dalam dapat dibuat tetap (seperti pola persegi) atau bergantung pada nilai variabel perulangan luar (seperti pola segitiga).
6. Fungsi print() dengan Parameter end
Secara bawaan, print() mengakhiri cetakan dengan pindah baris. Parameter end mengubah karakter penutup tersebut. Contohnya print("*", end=" ") mencetak tanda * diikuti spasi tanpa pindah baris, sehingga karakter berikutnya tercetak di sebelahnya. Sementara itu, print() tanpa argumen digunakan untuk pindah baris setelah satu baris pola selesai dicetak.
7. Masukan Data dengan input() dan Konversi Tipe dengan int()
Fungsi input() digunakan untuk menerima masukan dari pengguna dan selalu menghasilkan data bertipe teks (string). Agar dapat dipakai dalam perhitungan atau sebagai batas perulangan, data tersebut harus diubah ke bilangan bulat dengan fungsi int(), misalnya n = int(input("Masukkan nilai: ")).
8. Percabangan if, elif, dan else
Percabangan digunakan untuk mengambil keputusan berdasarkan suatu kondisi. Pernyataan if menjalankan blok kode jika kondisinya benar, elif memeriksa kondisi lain jika kondisi sebelumnya salah, dan else dijalankan jika semua kondisi sebelumnya tidak terpenuhi. Percabangan dapat diletakkan di dalam perulangan untuk menentukan karakter yang dicetak pada posisi tertentu, misalnya pada pola berbingkai.
9. Operator Perbandingan, Logika, dan Modulus<br>
-Operator perbandingan (==, !=, <, >, <=, >=) menghasilkan nilai benar atau salah.<br>
-Operator logika (and, or, not) menggabungkan beberapa kondisi.<br>
-Operator modulus (%) menghasilkan sisa hasil bagi. Operator ini dapat digunakan untuk menguji bilangan genap atau ganjil, misalnya n % 2 == 0 bernilai benar jika n genap.<br>
10. Variabel dan Aturan Penamaan
Variabel adalah tempat menyimpan data dalam program. Nama variabel pada Python bersifat case-sensitive (huruf besar dan kecil dibedakan), tidak boleh diawali angka, tidak boleh mengandung spasi, dan tidak boleh memakai kata kunci bawaan Python (reserved keywords) seperti for, if, atau while. Pada praktikum ini, setiap nama variabel diakhiri empat digit terakhir NIM sebagai penanda pengerjaan mandiri.
Komentar dan Indentasi
Komentar diawali tanda # dan diabaikan oleh interpreter, digunakan untuk menjelaskan kode. Indentasi pada Python bukan sekadar penataan tampilan, tetapi menentukan blok kode mana yang termasuk ke dalam suatu perulangan atau percabangan.

### 2.2.1 Program "A_2611532015"
**a. Program perulangan_for1_2611532015**


```python
ulang_2015 = int(input("Masukkan jumlah perulangan : "))
for i in range(ulang_2015) :
    # if i == 0:
    #     continue
    print(f"Perulangan ke-{i}")
```

**Penjelasan :**
1. ulang_2015 = int(input("Masukkan jumlah perulangan : "))<br>
Baris ini menampilkan teks "Masukkan jumlah perulangan : " lalu menunggu pengguna mengetik angka. Fungsi input() menghasilkan data bertipe teks, sehingga dibungkus int() untuk mengubahnya menjadi bilangan bulat. Hasilnya disimpan dalam variabel ulang_2015. Misalnya pengguna mengetik 3, maka ulang_2015 bernilai 3.
2. for i in range(ulang_2015) :<br>
Baris ini memulai perulangan for. Fungsi range(ulang_2015) menghasilkan rangkaian bilangan bulat dari 0 sampai ulang_2015 - 1. Untuk ulang_2015 = 3, rangkaiannya adalah 0, 1, 2. Variabel i menampung satu nilai dari rangkaian itu pada setiap putaran, sehingga perulangan berjalan sebanyak 3 kali.
3. print(f"Perulangan ke-{i}")<br>
Baris ini mencetak teks ke layar pada setiap putaran. Awalan f menandakan f-string, yaitu teks yang dapat menyisipkan nilai variabel di dalam kurung kurawal { }. Bagian {i} akan diganti dengan nilai i pada putaran tersebut. Karena tidak memakai end="", setiap cetakan berpindah baris.
4. Indentasi (penjorokan)<br>
Baris print(...) menjorok ke dalam di bawah for. Penjorokan ini menandakan bahwa baris tersebut termasuk isi perulangan dan ikut diulang. Baris tanpa penjorokan tidak ikut diulang.

**c. Program perulangan_for3_2611532015**

```python
ulang_2015 = int(input("Masukkan jumlah_2015 perulangan : "))
jumlah_2015 = 0
for i in range(1, ulang_2015+1):
    print(i, end=" ")
    jumlah_2015 += i

    if i < ulang_2015:
        print("+", end=" ")
    else:
        print("=", jumlah_2015, end=" " )

print()
print("Jumlah =", jumlah_2015)
```

range(1, ulang_2015+1) menghasilkan deret dari 1 sampai ulang_2015 (untuk input 5: 1, 2, 3, 4, 5). Variabel i menyimpan satu angka pada setiap putaran.
4. print(i, end=" ")<br>

5. jumlah_2015 += i<br>

**d. Program nested_for1_2611532015**


```python
batas_2015 = int(input("Masukkan nilai batas: "))
for line_2015 in range(1, batas_2015+1):
    for j_2015 in range(1, (-1 * line_2015 + batas_2015) + 1):
        print(".", end=" ")
    print(line_2015)

```

**Penjelasan:**

1. batas_2015 = int(input("Masukkan nilai batas: "))<br>
-input(...) menampilkan teks lalu menunggu pengguna mengetik.<br>
-int() mengubah input (string) menjadi bilangan bulat.<br>
-Nilainya disimpan di batas_2015, yang menentukan jumlah baris yang dicetak.<br>
2. for line_2015 in range(1, batas_2015+1):<br>
Ini perulangan luar (outer loop). range(1, batas_2015+1) menghasilkan deret 1 sampai batas_2015 (untuk input 5: 1, 2, 3, 4, 5). Variabel line_2015 menyimpan nomor baris yang sedang diproses, dan satu putaran perulangan luar menghasilkan satu baris output.
3. for j_2015 in range(1, (-1 * line_2015 + batas_2015) + 1):
-Ini perulangan dalam (inner loop), berada di dalam perulangan luar karena menjorok.<br>
-Batas akhirnya (-1 * line_2015 + batas_2015) + 1 sama dengan batas_2015 - line_2015 + 1, sehingga perulangan ini berjalan sebanyak batas_2015 - line_2015 kali.<br>
4. print(".", end=" ")<br>
Berada di dalam perulangan dalam, jadi dijalankan pada setiap putarannya. Baris ini mencetak titik diikuti spasi (bukan baris baru), sehingga titik-titik tampil sejajar dalam satu baris.
5. print(line_2015)<br>
Menjorok sejajar dengan for j_2015, jadi berada di luar perulangan dalam tetapi masih di dalam perulangan luar. Baris ini dijalankan sekali setiap baris, setelah semua titik tercetak. Ia mencetak nomor baris lalu pindah ke baris baru (bawaan print).

**Penjelasan:**

3. for j_2015 in range(1, batas_2015 + 1):<br>
-Ini perulangan dalam (inner loop), berada di dalam perulangan luar karena menjorok.<br>
-Berbeda dengan kode sebelumnya, batasnya tetap (tidak bergantung pada i_2015), sehingga selalu berjalan sebanyak batas_2015 kali pada setiap baris.
4. print("*", end=" ")<br>
Berada di dalam perulangan dalam, jadi dijalankan pada setiap putarannya. Baris ini mencetak bintang diikuti spasi (bukan baris baru), sehingga bintang-bintang tampil sejajar dalam satu baris.

Menjorok sejajar dengan for j_2015, jadi berada di luar perulangan dalam tetapi masih di dalam perulangan luar. Baris ini dijalankan sekali setiap baris, setelah semua bintang tercetak, untuk pindah ke baris baru.

<p align="justify">Berdasarkan praktikum yang telah dilaksanakan, dapat disimpulkan bahwa perulangan for merupakan konsep fundamental dalam pemrograman yang memungkinkan suatu instruksi dijalankan berulang kali tanpa harus ditulis berulang-ulang. Pada Python, jumlah dan arah perulangan dikendalikan melalui fungsi range(): range(stop) untuk perulangan yang dimulai dari 0, range(start, stop) untuk perulangan dengan nilai awal tertentu seperti dari 1 sampai N, serta range(start, stop, step) untuk mengatur kenaikan atau penurunan nilai. Untuk kebutuhan dua dimensi, digunakan perulangan bersarang (nested loop), di mana perulangan luar mengatur baris dan perulangan dalam mengatur kolom. Konsep ini diterapkan pada berbagai pola, mulai dari deretan titik dengan nomor baris yang jumlahnya berkurang, persegi berisi tanda bintang, tabel angka hasil penjumlahan indeks baris dan kolom, hingga pola berbingkai yang mengombinasikan nested loop dengan percabangan if-elif-else untuk menentukan karakter pada posisi tertentu. Fungsi print() dengan parameter end digunakan untuk menjaga karakter tetap dalam satu baris, sedangkan print() tanpa argumen digunakan untuk berpindah ke baris berikutnya.</p>

<p align="justify">Pemahaman yang tepat mengenai batas range() dan urutan eksekusi perulangan sangat memengaruhi kebenaran hasil program, karena nilai batas tidak ikut dihitung dan perulangan dalam selalu diselesaikan sepenuhnya pada setiap putaran perulangan luar. Ketelitian dalam menentukan jumlah ulangan dan penjorokan (indentasi) juga menentukan bentuk pola yang dihasilkan. Melalui praktikum ini, penulis berhasil memahami cara kerja perulangan for dan nested loop serta mampu mengimplementasikan, menganalisis, dan menelusuri alur program untuk membentuk berbagai pola yang lebih kompleks..</p>
