## 2.1 Dasar Teori

1. Struktur Percabangan (Conditional Statement)<br>
Struktur percabangan adalah struktur kontrol dalam pemrograman yang memungkinkan program untuk memilih dan mengeksekusi blok kode tertentu berdasarkan hasil evaluasi suatu kondisi (bernilai benar/True atau salah/False). Struktur ini menjadi dasar bagi program dalam mengambil keputusan sesuai dengan situasi yang dihadapi.
2. Pernyataan if<br>
Pernyataan if merupakan bentuk percabangan paling sederhana yang digunakan untuk mengeksekusi suatu blok kode hanya jika kondisi yang diberikan bernilai benar (True). Jika kondisi bernilai salah (False), maka blok kode di dalamnya akan dilewati.
3. Pernyataan if-else<br>
Pernyataan if-else digunakan untuk menangani dua kemungkinan hasil, yaitu satu blok kode akan dijalankan jika kondisi bernilai benar, dan blok kode lain (di bagian else) akan dijalankan jika kondisi bernilai salah.
4. Pernyataan if-elif-else<br>
Pernyataan if-elif-else digunakan ketika terdapat lebih dari dua kemungkinan kondisi yang perlu dievaluasi secara berurutan. Python akan memeriksa setiap kondisi dari atas ke bawah, dan hanya blok kode dari kondisi pertama yang bernilai benar yang akan dieksekusi.
5. Multi-If (Nested If)<br>
Multi-if atau nested if adalah struktur percabangan yang diletakkan di dalam struktur percabangan lainnya (percabangan bersarang). Struktur ini digunakan ketika suatu kondisi hanya perlu dievaluasi setelah kondisi lain terpenuhi terlebih dahulu, sehingga cocok untuk permasalahan dengan tingkat kondisi yang lebih kompleks atau bertingkat.
6. Pernyataan match-case<br>
match-case adalah fitur pattern matching yang diperkenalkan pada Python versi 3.10 sebagai alternatif struktur percabangan, khususnya untuk kasus dengan banyak kemungkinan nilai yang dibandingkan terhadap satu variabel. Struktur ini serupa dengan switch-case pada bahasa pemrograman lain dan menawarkan penulisan kode yang lebih ringkas serta mudah dibaca dibandingkan if-elif-else yang panjang.

**d. Program multi_if2_2611532015**


```python
#input dari user
total_belanja_2015 = float(input("Masukkan total belanja (Rp) : "))

# input status member 
input_member_2015 = input("Apakah anda member ?(y/t) : ")
is_member_2015 = input_member_2015 in ["y", 'ya']

#mengecek status kode promo
input_promo_2015 = input("Apakah kode promo valid? (y/t) : ").strip().lower()
kode_promo_valid_2015 = input_promo_2015 in ["y", 'ya']

total_diskon_persen = 0
# multi if terpisah : setiap kondisi diperiksa secara independen
# diskon bisa ditumpuk (akumulasi) jika memnuhi beberapa syarat sekaligus

if total_belanja_2015 > 1000000 :
    total_diskon_persen += 10 # diskon belanja besar

if is_member_2015 :
    total_diskon_persen += 5 # diskon member

if kode_promo_valid_2015 : 
    total_diskon_persen += 15 # diskon voucher

#menghitung nominal diskon dan total bayar
nominal_diskon_2015 = total_belanja_2015 * (total_diskon_persen/100)
total_bayar_2015 = total_belanja_2015 - nominal_diskon_2015

#output hasil
print("\n===RINCIAN PEMBAYARAN===")
print(f"Total diskon   : {total_diskon_persen}% (Rp {nominal_diskon_2015:,.0f})")
print(f"Total bayar    : Rp {total_bayar_2015:,.0f}")

print(f"total diskon yang anda dapatkan : {total_diskon_persen}%")
```

**Penjelasan:**

1. Program meminta pengguna memasukkan total belanja dalam bentuk float melalui input(), dan menyimpannya ke variabel total_belanja_2015.
2. Program meminta status keanggotaan (member) pengguna melalui input(), kemudian menggunakan operator in untuk memeriksa apakah jawaban pengguna berada di dalam list ["y", "ya"]. Hasil pemeriksaan ini (True/False) langsung disimpan ke variabel is_member_2015.
3. Program meminta status kode promo dengan cara serupa, namun input terlebih dahulu diproses menggunakan .strip() untuk menghapus spasi berlebih dan .lower() untuk menyamakan huruf menjadi kecil semua, sebelum diperiksa apakah termasuk dalam ["y", "ya"] dan disimpan ke variabel kode_promo_valid_2015.
4. Variabel total_diskon_persen diinisialisasi dengan nilai 0 sebagai penampung akumulasi persentase diskon yang akan bertambah sesuai kondisi yang terpenuhi.
5. Program menggunakan tiga pernyataan if yang berdiri sendiri (multi-if), bukan if-elif-else, sehingga setiap kondisi diperiksa secara independen dan diskonnya bisa saling ditumpuk (akumulasi) jika beberapa syarat terpenuhi sekaligus:<br>
-Jika total_belanja_2015 > 1000000, maka total_diskon_persen ditambah 10 (diskon belanja besar).<br>
-Jika is_member_2015 bernilai True, maka total_diskon_persen ditambah 5 (diskon member).<br>
-Jika kode_promo_valid_2015 bernilai True, maka total_diskon_persen ditambah 15 (diskon promo).<br>
6. Setelah seluruh kondisi dievaluasi, program menghitung nominal_diskon_2015 dengan mengalikan total_belanja_2015 dengan persentase total diskon yang telah diakumulasi (total_diskon_persen/100), sehingga diperoleh nominal diskon dalam bentuk rupiah. Kemudian total_bayar_2015 dihitung dengan mengurangkan nominal diskon tersebut dari total_belanja_2015.
7. Terakhir, program mencetak rincian pembayaran, meliputi total diskon dalam bentuk persentase beserta nominalnya (menggunakan format :,.0f agar angka tampil dengan pemisah ribuan dan tanpa desimal), total yang harus dibayar, dan persentase diskon yang didapatkan pengguna.

**g. Program match_case1_2611532015**

```python
#buat program untuk match case
bulan = int(input("Masukkan angka bulan (1-12) : "))

match bulan :
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("Sepetember")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")

```

**Penjelasan: **
1. Program meminta pengguna memasukkan angka bulan (1–12) melalui input() yang dikonversi ke tipe data int, lalu disimpan ke variabel bulan.
2. Program menggunakan struktur percabangan match-case, yaitu fitur pattern matching yang tersedia mulai Python 3.10, sebagai alternatif dari if-elif-else yang panjang untuk kasus dengan banyak kemungkinan nilai.
3. Variabel bulan yang dievaluasi pada match bulan akan dicocokkan dengan setiap case secara berurutan dari atas ke bawah. Setiap case merepresentasikan satu kemungkinan nilai angka bulan (1 sampai 12) beserta nama bulannya masing-masing dalam Bahasa Indonesia.
4. Ketika ditemukan case yang nilainya cocok dengan bulan (misalnya bulan bernilai 5), maka blok kode pada case tersebut akan dieksekusi (mencetak "Mei"), dan pencocokan langsung berhenti tanpa memeriksa case lain di bawahnya.
5. Apabila angka yang dimasukkan pengguna tidak cocok dengan case mana pun (misalnya bernilai 0, 13, atau negatif), maka case _ akan dijalankan sebagai kondisi default (wildcard), yang berfungsi serupa dengan else pada struktur percabangan lainnya, dan mencetak "Angka tidak valid".
6. Struktur match-case ini memberikan penulisan kode yang lebih ringkas dan mudah dibaca dibandingkan penggunaan if-elif-else sebanyak 12 kondisi, terutama untuk kasus pencocokan nilai tunggal seperti pemetaan angka bulan ke nama bulan.

**Kode Program:**

```python
print("===SISTEM LOKET ALPRO ADVENTURE PARK===")
nama_2015 = input("\nMasukkan nama pengunjung\t\t : ")
umur_2015 = int(input("Masukkan umur pengunjung\t\t : "))
sim_2015 = input("Apakah anda sudah punya SIM C? (y/t)\t : ").strip().lower()[0]


print("\n\t\t====pilihan paket wahana====".upper())
print("\n\t1. Wahana Safari Rimba (Rp. 50.000)")
print("\t2. Wahana Arung Jeram (Rp. 75.000)")
print("\t3. Wahana Motor ATV Ekstrim (Rp. 120.000)")
print("\t4. Wahana Roller Coaster Kilat (Rp. 100.000)")
print("\t5. Wahana All-Access VIP (Rp. 220.000)")
paket_2015 = int(input("\n\tMasukkan nomor paket yang anda inginkan(1-5) : "))

match paket_2015:
    case 1:
        nama_wahana_2015 = "Wahana Safari Rimba"
        harga_satuan_2015 = 50000
    case 2:
        nama_wahana_2015 = "Wahana Arung Jeram"
        harga_satuan_2015 = 75000
    case 3:
        nama_wahana_2015 = "Wahana Motor ATV Ekstrim"
        harga_satuan_2015 = 120000
    case 4:
        nama_wahana_2015 = "Wahana Roller Coaster Kilat"
        harga_satuan_2015 = 100000
    case 5:
        nama_wahana_2015 = "Wahana All-Access VIP"
        harga_satuan_2015 = 220000
    case _:
        print("Paket Wahana tidak valid!")
        exit()

jumlah_tiket_2015 = int(input("\tMasukkan jumlah tiket anda\t\t : "))
# if tunggal
if jumlah_tiket_2015 <= 0:
    print("\n==TIKET TIDAK VALID==")
    exit()

is_member_2015 = input("\tApakah anda member? (y/t)\t\t : ").strip().lower()[0]
kode_promo_valid_2015 = input("\tApakah kode promo valid? (y/t)\t : ").strip().lower()[0]

print("\n\t===KELAYAKAN PENGENDARA WAHANA===")
if paket_2015 == 3:
    if umur_2015 >= 17 and sim_2015 == "y":
        print("\tStatus Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri")
    elif umur_2015 >= 17 and sim_2015 != "y":
        print("\tStatus Akses: Anda sudah dewasa tetapi tidak boleh mengendarai ATV sendiri")
    elif umur_2015 < 17 and sim_2015 == "y":
        print("\tStatus Akses: Identitas tidak valid: Anda belum cukup umur untuk memiliki SIM")
    else:
        print("\tStatus Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV")

# untuk selain paket 3
if umur_2015 >= 10:
    print("\tStatus Akses: Anda memenuhi syarat minimum untuk wahana ini")
else:
    print("\tStatus Akses: Anda belum memenuhi syarat minimum untuk wahana ini")

# hitung diskon
subtotal_2015 = harga_satuan_2015 * jumlah_tiket_2015
total_diskon_persen = 0

if subtotal_2015 >= 200000:
    total_diskon_persen += 10  # diskon belanja besar
if is_member_2015 in ['y', 'ya']:
    total_diskon_persen += 5   # diskon member
if kode_promo_valid_2015 in ["y", "ya"]:
    total_diskon_persen += 15  # diskon voucher promo
if jumlah_tiket_2015 >= 5:
    total_diskon_persen += 5   # diskon tambahan rombongan

nominal_diskon_2015 = subtotal_2015 * (total_diskon_persen / 100)
total_bayar_2015 = subtotal_2015 - nominal_diskon_2015

print("\n\t===RINCIAN PEMBAYARAN===")
print(f"\tSubtotal Belanja : Rp. {subtotal_2015:,.0f}")
print(f"\tTotal Diskon\t : {total_diskon_persen}%")
print(f"\tTotal Bayar\t : Rp. {total_bayar_2015:,.0f}")

if total_bayar_2015 >= 300000:
    print("\n\tCatatan Layanan: Selamat! Anda berhak mendapatkan Souvenir Gratis")
else:
    print("\tCatatan Layanan: Terima kasih telah berkunjung")

print("\n\t\t====Program Selesai====")
```

1. Program mencetak judul sistem, kemudian meminta input data pengunjung: nama (nama_2015), umur dalam bentuk int (umur_2015), dan status kepemilikan SIM C yang diproses dengan .strip(), .lower(), dan [0] agar hanya karakter pertama yang diambil, lalu disimpan ke sim_2015.
2. Program menampilkan daftar 5 pilihan paket wahana beserta harganya, kemudian meminta pengguna memasukkan nomor paket (1–5) ke variabel paket_2015.
3. Program menggunakan struktur match-case untuk mencocokkan paket_2015 dengan setiap case. Setiap case yang cocok akan mengisi nama_wahana_2015 dan harga_satuan_2015 sesuai paket yang dipilih. Jika nomor paket tidak sesuai dengan case mana pun, case _ akan dijalankan, mencetak pesan "Paket Wahana tidak valid!", lalu program dihentikan menggunakan exit() agar tidak melanjutkan proses dengan data yang tidak valid.
4. Program meminta input jumlah tiket (jumlah_tiket_2015), lalu menggunakan if tunggal untuk memeriksa apakah jumlahnya kurang dari atau sama dengan 0. Jika benar, program mencetak "TIKET TIDAK VALID" dan langsung dihentikan dengan exit(), sehingga program tidak melanjutkan proses dengan jumlah tiket yang tidak masuk akal.
5. Program meminta input status member (is_member_2015) dan status kode promo (kode_promo_valid_2015) dengan cara input yang sama seperti sebelumnya.
6. Program menggunakan nested if untuk memeriksa kelayakan pengendara khusus paket ATV: if luar memeriksa paket_2015 == 3 terlebih dahulu, dan hanya jika kondisi ini benar, struktur if-elif-else di dalamnya akan dievaluasi untuk memeriksa kombinasi umur dan kepemilikan SIM. Dengan struktur ini, blok pesan kelayakan ATV hanya akan tercetak jika pengunjung benar-benar memilih paket 3, dan tidak akan memengaruhi pengunjung yang memilih paket lain.
7. Program menggunakan if-else terpisah untuk memeriksa syarat umur minimum wahana secara umum (umur_2015 >= 10), yang berlaku untuk seluruh jenis paket, dan mencetak status apakah pengunjung memenuhi syarat tersebut.
8. Program menghitung subtotal_2015 dengan mengalikan harga_satuan_2015 dengan jumlah_tiket_2015, lalu menginisialisasi total_diskon_persen dengan nilai 0.
9. Program menggunakan empat pernyataan if independen (multi-if) untuk mengakumulasi diskon, sehingga beberapa diskon bisa digabung sekaligus jika syaratnya terpenuhi:<br>
-Subtotal ≥ Rp200.000 → tambah diskon 10% (belanja besar).<br>
-Status member ("y"/"ya") → tambah diskon 5%.<br>
-Kode promo valid ("y"/"ya") → tambah diskon 15%.<br>
-Jumlah tiket ≥ 5 → tambah diskon 5% (rombongan).
10. Program menghitung nominal_diskon_2015 dengan mengalikan subtotal_2015 dengan persentase total diskon (total_diskon_persen/100), kemudian total_bayar_2015 dihitung dengan mengurangkan nominal diskon dari subtotal.
11. Program mencetak rincian pembayaran (subtotal, total diskon, dan total bayar) menggunakan format :,.0f agar angka tampil dengan pemisah ribuan tanpa desimal.
12. Terakhir, program menggunakan if-else untuk memeriksa apakah total_bayar_2015 mencapai Rp300.000 ke atas. Jika iya, pengunjung mendapat pesan bonus souvenir gratis; jika tidak, program mencetak ucapan terima kasih. Program ditutup dengan pesan "Program Selesai".
