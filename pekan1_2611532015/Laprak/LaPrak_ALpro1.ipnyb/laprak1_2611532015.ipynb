## 2.1 Dasar Teori

1. Python.  Python adalah bahasa pemrograman tingkat tinggi yang bersifat interpreted dan general-purpose. Bahasa ini dikenal dengan sintaks yang sederhana, mudah dipahami, serta memiliki pustaka standar yang luas. Python banyak digunakan dalam bidang analisis data, pengembangan web, kecerdasan buatan, dan otomasi.

2. Visual Studio Code (VS Code).VS Code merupakan source-code editor yang dikembangkan oleh Microsoft. Editor ini mendukung berbagai bahasa pemrograman, termasuk Python, serta dilengkapi fitur seperti extensions, debugging, dan integrasi dengan Git. VS Code menjadi salah satu editor populer karena ringan, fleksibel, dan ramah pengguna.

3. Git dan GitHub.Git adalah sistem kontrol versi (version control system) yang digunakan untuk melacak perubahan pada kode. GitHub adalah platform berbasis web yang menyediakan layanan penyimpanan repositori Git serta mendukung kolaborasi antar pengguna. Dengan GitHub, pengembang dapat mengelola proyek secara terstruktur, berbagi kode, dan bekerja dalam tim.
