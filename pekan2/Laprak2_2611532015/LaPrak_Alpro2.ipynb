## 2.1 Dasar Teori

1. Numerik<br>
Tipe data numerik digunakan untuk merepresentasikan bilangan bulat maupun pecahan dalam pemrograman. Komputasi numerik menjadi fondasi dalam berbagai aplikasi IT seperti machine learning, grafika komputer, dan simulasi sistem kompleks. Representasi bilangan dan algoritma numerik membantu komputer melakukan operasi aritmatika dengan presisi tertentu.

2. String. <br>
String adalah kumpulan karakter yang digunakan untuk menyimpan dan memanipulasi teks. Dalam bahasa Python, string direpresentasikan dengan tanda kutip tunggal ('...') atau ganda ("...") dan dapat berisi huruf, angka, simbol, maupun karakter khusus.
3. Input. <br>
Input adalah mekanisme untuk menerima data dari pengguna agar program dapat berinteraksi secara dinamis. Fungsi input memungkinkan nilai variabel ditentukan saat program berjalan, bukan hanya saat kompilasi. Hal ini penting untuk membangun aplikasi interaktif.
4. Boolean. <br>
Boolean adalah tipe data logika yang hanya memiliki dua nilai: true dan false. Konsep ini diperkenalkan oleh George Boole dan menjadi dasar dalam pengambilan keputusan (percabangan, loop, kontrol alur). Boolean juga digunakan dalam desain rangkaian digital dan algoritma logika.
5. Konstanta. <br>
Konstanta adalah nilai tetap yang tidak dapat diubah selama eksekusi program. Penggunaan konstanta menjaga konsistensi dan memudahkan pemeliharaan kode, misalnya mendefinisikan nilai π atau batas maksimum. Konstanta dapat ditulis dengan const atau #define tergantung bahasa pemrograman.

**b. Program Numerik**

```python
a_2015 = 100
print("The type of variabel having value", a_2015,"is", type(a_2015))
b_2015 = 10.2345
print("The type of variabel having value", b_2015,"is", type(b_2015))
c_2015 = 100+3j
print("The type of variabel having value", c_2015,"is", type(c_2015))

```

**Penjelasan:**

1. Baris 1 → Variabel a_2015 menyimpan nilai integer (bilangan bulat).
2. Baris 2 → print() menampilkan nilai variabel a_2015 sekaligus menunjukkan tipe datanya dengan fungsi type().
3. Baris 3 → Variabel b_2015 menyimpan nilai float (bilangan desimal).
4. Baris 4 → print() menampilkan nilai variabel b_2015 sekaligus menunjukkan tipe datanya dengan fungsi type().
5. Baris 5 → Variabel c_2015 menyimpan nilai complex (bilangan kompleks dengan bagian imajiner).
6. Baris 6 → print() menampilkan nilai variabel c_2015 sekaligus menunjukkan tipe datanya dengan fungsi type().

**c. Program Boolean.**

```python
is_lulus_2015 = True
is_cumloude_2015 = True
nilai_2015 = 90
batas_lulus_2015 = 75

status_kelulusan_2015 = nilai_2015 >= batas_lulus_2015 #Hasilnya akan lulus 
print("=== Check Kelulusan ===")
print("Nilai", nilai_2015)
print("Apakah Lulus?:", status_kelulusan_2015)
if is_lulus_2015 and is_cumloude_2015:
    print("Selamat, Anda lulus dengan predikat CumLaude!")
```

**Penjelasan:**

1. Baris 1–2 → Variabel is_lulus_2015 dan is_cumloude_2015 menyimpan nilai boolean (True) untuk menandai status kelulusan dan predikat CumLaude.
2. Baris 3–4 → Variabel nilai_2015 dan batas_lulus_2015 menyimpan nilai numerik (integer) untuk skor ujian dan batas minimal kelulusan.
3. Baris 6 → Variabel status_kelulusan_2015 menyimpan hasil perbandingan logika nilai_2015 >= batas_lulus_2015, menghasilkan nilai boolean sesuai kondisi.
4. Baris 7–9 → Fungsi print() digunakan untuk menampilkan teks keterangan dan nilai variabel ke layar.
5. Baris 10–11 → Struktur kontrol if mengecek kondisi gabungan is_lulus_2015 and is_cumloude_2015. Jika keduanya bernilai True, maka program menampilkan pesan “Selamat, Anda lulus dengan predikat CumLaude!”.

**d. Program Input**

```python
panjang_2015 = input("Masukkan nilai panjang: ")
lebar_2015 = input('Masukkan nilai lebar: ')
luas_2015 = int(panjang_2015) * int(lebar_2015)
print("Luas:", luas_2015)
```

**Penjelasan:**

1. Baris 1 → panjang_2015 = input("Masukkan nilai panjang: ") berfungsi untuk menerima input dari pengguna berupa teks (string) dan menyimpannya ke variabel panjang_2015.
2. Baris 2 → lebar_2015 = input('Masukkan nilai lebar: ') berfungsi untuk menerima input dari pengguna berupa teks (string) dan menyimpannya ke variabel lebar_2015.
3. Baris 3 → luas_2015 = int(panjang_2015) * int(lebar_2015) mengubah input string menjadi integer dengan fungsi int(), lalu menghitung perkalian panjang × lebar untuk mendapatkan nilai luas.
4. Baris 3 → luas_2015 = int(panjang_2015) * int(lebar_2015) mengubah input string menjadi integer dengan fungsi int(), lalu menghitung perkalian panjang × lebar untuk mendapatkan nilai luas.

**e. Program Konstanta**

```python
from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_2015 = float(input('Masukkan nilai jari-jari: '))
luas_2015 = PI * jari_2015 * jari_2015
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2015, luas_2015))
```

**Penjelasan:**

1. Baris 1 → from typing import Final digunakan untuk mengimpor anotasi Final dari modul typing, yang menandai bahwa sebuah variabel bersifat konstan (tidak boleh diubah).
2. Baris 2 → PI: Final = 3.14 mendefinisikan konstanta PI dengan nilai 3.14.
3. Baris 3 → print("pi: %f" % (PI)) menampilkan nilai konstanta PI dalam format float.
4. Baris 4 → jari_2015 = float(input('Masukkan nilai jari-jari: ')) menerima input dari pengguna, mengubahnya menjadi tipe data float, lalu menyimpannya ke variabel jari_2015.
5. Baris 5 → luas_2015 = PI * jari_2015 * jari_2015 menghitung luas lingkaran dengan rumus π × r × r menggunakan nilai konstanta PI dan variabel jari_2015.
6. Baris 6 → print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2015, luas_2015)) menampilkan hasil perhitungan luas lingkaran dengan format dua angka di belakang koma.

# Tugas Praktikum

**Studi Kasus: Sistem Biodata dan Validasi Kelulusan Praktikan**
<br>Anda diminta untuk merancang sebuah program konsol sederhana yang mencatat data
identitas praktikan baru dan melakukan verifikasi status kelulusan awal praktikum secara
otomatis. Seluruh elemen materi Pekan 2 wajib diintegrasikan dalam satu alur program yang
utuh.
<br>


**Kode Program:**

```python
# Sistem Biodata dan Validasi Kelulusan Praktikan

# Konstanta batas kelulusan
batas_kelulusan_2015 = 75.0

# Input biodata praktikan
nama_2015 = input("Masukkan Nama: ")
nim_2015 = input("Masukkan NIM: ")
jurusan_2015 = input("Masukkan Jurusan: ")
fakultas_2015 = input("Masukkan Fakultas: ")
universitas_2015 = input("Masukkan Universitas: ")

# Input nilai awal praktikum
nilai_2015 = float(input("Masukkan Nilai Awal Praktikum: "))

# Output biodata
print("\n=== BIODATA PRAKTIKAN ===")
print("Nama:", nama_2015)
print("NIM:", nim_2015)
print("Jurusan:", jurusan_2015)
print("Fakultas:", fakultas_2015)
print("Universitas:", universitas_2015)

# Validasi kelulusan
print("\n=== HASIL VALIDASI ===")
if nilai_2015 >= batas_kelulusan_2015:
    print("Status: LULUS ")
else:
    print("Status: TIDAK LULUS ")
```

**Penjelasan:**

1. Konstanta batas kelulusan
    batas_kelulusan_2015 = 80.0 <BR>
    Menentukan ambang batas kelulusan yaitu 80.0. <BR>
    Tipe data float digunakan agar bisa menerima angka desimal.<BR>

2. Input biodata praktikan
    input() digunakan untuk meminta pengguna memasukkan:<br>
    Nama<br>
    NIM<br>
    Jurusan<br>
    Fakultas<br>
    Universitas<br>

3. Input nilai awal praktikum
    nilai_2015 = float(input("Masukkan Nilai Awal Praktikum: "))<br>
    Nilai praktikum dimasukkan oleh pengguna.<br>
    Fungsi float() memastikan nilai berupa angka desimal, bukan teks.<br>

4. Output biodata
    <br>Program menampilkan kembali biodata yang sudah diinput:<br>
    Nama<br>
    NIM<br>
    Jurusan<br>
    Fakultas<br>
    Universitas<br>
    Bagian ini berfungsi sebagai konfirmasi data.

5. Validasi kelulusan<br>
    Program membandingkan nilai praktikum dengan batas_kelulusan_2015:<br>
    Jika nilai ≥ 80 → Status: LULUS<br>
    Jika nilai < 80 → Status: TIDAK LULUS
